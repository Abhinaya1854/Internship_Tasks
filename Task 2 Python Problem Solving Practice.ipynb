```python
# Q1. Find the Shortest Word in a Sentence
# Observe: Input is a sentence,output is the shortest word.
# Understand: Find the word with the minimum length.
# Break: Split the sentence and compare word lengths.
# Logic: Start with the first word and update when a shorter word is found.
# Code: Use split(), len(), and a for loop.
s = "Python makes coding easy"
words = s.split()

shortest = words[0]

for word in words:
    if len(word) < len(shortest):
        shortest = word

print(shortest)
```

```python
# Q2. Find the Largest Number Without max()
# Observe: Input is a list of numbers, output is the largest number.
# Understand: Find the maximum number without using max().
# Break: Start with the first number and compare all numbers.
# Logic: Update the largest number whenever a bigger number appears.
# Code: Use a for loop and an if condition.

nums = [12, 45, 7, 89, 23]
largest = nums[0]

for num in nums:
    if num > largest:
        largest = num

print(largest)
```

```python
# Q3. Find the Smallest Number Without min()
# Observe: Input is a sentence; output is the first longest word.
# Understand: Find the longest word while preserving the first word in a tie.
# Break: Split the sentence and compare word lengths.
# Logic: Update the result only when a strictly longer word is found.
# Code: Use split(), len(), and a for loop.

nums = [12, 45, 7, 89, 23]
smallest = nums[0]

for num in nums:
    if num < smallest:
        smallest = num

print(smallest)
```

```python
# Q4. Find the First Longest Word
# Observe: Input is a sentence,output is the first longest word.
# Understand: Find the longest word while preserving the first word in a tie.
# Break: Split the sentence and compare word lengths.
# Logic: Update the result only when a strictly longer word is found.
# Code: Use split(), len(), and a for loop.

sen = "Python makes coding useful"
words = sen.split()
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print(longest)
```

```python
# Q5. Find the Word with the Maximum Characters
# Observe: Input is a sentence,output is the word with the most characters.
# Understand: Identify the word having the greatest length.
# Break: Split the sentence and compare each word's length.
# Logic: Store the longest word and update it when a longer word appears.
# Code: Use split(), len(), and a for loop.

sentence = "Data engineering requires problem solving"
words = sentence.split()
result = words[0]
for word in words:
    if len(word) > len(result):
        result = word
print(result)
```

```python
# Q6. Count Vowels
# Observe: Input is a string; output is the total number of vowels.
# Understand: Count both uppercase and lowercase vowels.
# Break: Check each character against the vowels aeiou.
# Logic: Increase the count whenever a vowel is found.
# Code: Use a for loop, lower(), and an if condition.

text = "Education"
count = 0
for char in text:
    if char.lower() in "aeiou":
        count += 1
print(count)
```

```python
# Q7. Count Consonants
# Observe: Input is a string,output is the consonant count.
# Understand: Count letters that are not vowels, ignoring digits and symbols.
# Break: Check each character to determine whether it is a letter and a consonant.
# Logic: Count alphabetic characters that are not vowels.
# Code: Use isalpha(), lower(), and an if condition.

text = "Python 123!"
count = 0
for char in text:
    if char.isalpha() and char.lower() not in "aeiou":
        count += 1
print(count)
```

```python
# Q8. Find the Most Frequent Character
# Observe: Input is a string,output is its most frequent character.
# Understand: Find the character with the highest frequency, ignoring spaces.
# Break: Count each character and compare its frequency.
# Logic: Keep the first character with the highest frequency.
# Code: Use count(), a for loop, and an if condition.

text = "banana".replace(" ", "")
most_frequent = text[0]
for char in text:
    if text.count(char) > text.count(most_frequent):
        most_frequent = char
print(most_frequent)
```

```python
# Q9. Find the Least Frequent Character
# Observe: Input is a string,output is its least frequent character.
# Understand: Find the character with the lowest frequency, ignoring spaces.
# Break: Count each character and compare its frequency.
# Logic: Keep the first character with the lowest frequency.
# Code: Use count(), a for loop, and an if condition

text = "success".replace(" ", "")
least_frequent = text[0]
for char in text:
    if text.count(char) < text.count(least_frequent):
        least_frequent = char
print(least_frequent)
```

```python
# Q10. Find All Repeated Characters
# Observe: Input is a string,output is a list of repeated characters.
# Understand: Find repeated characters once, ignoring spaces.
# Break: Check the frequency of each character and whether it is already listed.
# Logic: Add a character if it appears more than once and is not in the result.
# Code: Use count(), a list, and a for loop.

text = "programming"
result = []
for char in text:
    if char != " " and text.count(char) > 1 and char not in result:
        result.append(char)
print(result)
```

```python
# Q11. Dictionary of Words and First Characters
# Observe: Input is a sentence; output is a dictionary of words and first characters.
# Understand: Store each word as a key and its first character as the value.
# Break: Split the sentence and access each word's first character.
# Logic: Add each word and its first character to a dictionary.
# Code: Use split(), a dictionary, and a for loop.

sentence = "Python Data Engineering"
result = {}
for word in sentence.split():
    result[word] = word[0]
print(result)
```

```python
# Q12. Dictionary of Words and Last Characters
# Observe: Input is a sentence,output is a dictionary of words and last characters.
# Understand: Store each word as a key and its last character as the value.
# Break: Split the sentence and access each word's last character.
# Logic: Add each word and its last character to a dictionary.
# Code: Use split(), word[-1], and a for loop.

sentence = "Python Data Engineering"
result = {}
for word in sentence.split():
    result[word] = word[-1]
print(result)
```

```python
# Q13. Dictionary of Words and Uppercase Versions
# Observe: Input is a sentence,output is a dictionary of words and uppercase versions.
# Understand: Map each word to its uppercase form.
# Break: Split the sentence and convert each word to uppercase.
# Logic: Store each original word as a key and its uppercase version as the value.
# Code: Use split(), upper(), and a dictionary.

sentence = "data science python"
result = {}
for word in sentence.split():
    result[word] = word.upper()
print(result)
```

```python
# Q14. Dictionary of Numbers and Their Squares
# Observe: Input is a list of numbers,output is a dictionary of squares.
# Understand: Store each number as a key and its square as the value.
# Break: Visit each number and calculate its square.
# Logic: Assign each number and its squared value to the dictionary.
# Code: Use a for loop, a dictionary, and num ** 2.

numbers = [2, 3, 4, 5]
result = {}
for num in numbers:
    result[num] = num ** 2
print(result)
```

```python
# Q15. Dictionary of Even/Odd Status
# Observe: Input is a list of integers,output is a dictionary of even/odd labels.
# Understand: Classify each number as "Even" or "Odd".
# Break: Divide each number by two and check the remainder.
# Logic: If the remainder is zero, store "Even"; otherwise, store "Odd".
# Code: Use %, an if-else condition, and a dictionary.

numbers = [1, 2, 3, 4, 5]
result = {}
for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"
print(result)
```

```python
# Q16. Find the Most Frequent Word
# Observe: Input is a sentence,output is the most frequent word.
# Understand: Find the most frequent word, treating uppercase and lowercase as equal.
# Break: Convert words to lowercase and count their occurrences.
# Logic: Keep the first word with the highest frequency.
# Code: Use lower(), split(), count(), and a for loop.

sentence = "cat dog cat bird dog"
words = sentence.lower().split()
most_frequent = words[0]
for word in words:
    if words.count(word) > words.count(most_frequent):
        most_frequent = word
print(most_frequent)
```

```python
# Q17. Find All Repeated Words
# Observe: Input is a sentence,output is a list of repeated words.
# Understand: Find repeated words once, ignoring uppercase and lowercase differences.
# Break: Convert words to lowercase and count their occurrences.
# Logic: Add each repeated word only once, in its first-appearance order.
# Code: Use lower(), split(), count(), and a list.

sentence = "data science uses data and science tools"
words = sentence.lower().split()
result = []
for word in words:
    if words.count(word) > 1 and word not in result:
        result.append(word)
print(result)
```

```python
# Q18. Find the First Word That Appears Only Once
# Observe: Input is a sentence,output is the first non-repeated word.
# Understand: Find the first word occurring exactly once, ignoring case.
# Break: Convert words to lowercase and count each word.
# Logic: Print the first word whose frequency is one and stop searching.
# Code: Use lower(), split(), count(), and break.

sentence = "red blue red green"
words = sentence.lower().split()
for word in words:
    if words.count(word) == 1:
        print(word)
        break
```

```python
# Q19. Count Words Longer Than Four Characters
# Observe: Input is a sentence,output is the number of words longer than four characters.
# Understand: Count words whose length is greater than four.
# Break: Split the sentence and check each word's length.
# Logic: Increase the count whenever len(word) > 4.
# Code: Use split(), len(), a counter, and a for loop.

sentence = "I love data science"
words = sentence.split()
count = 0
for word in words:
    if len(word) > 4:
        count += 1
print(count)
```

```python
# Q20. Find the Longest Word That Appears Only Once
# Observe: Input is a sentence,output is the longest word appearing once.
# Understand: Find the longest unique word, ignoring uppercase and lowercase differences.
# Break: Convert words to lowercase, count occurrences, and compare lengths.
# Logic: Consider only words occurring once and keep the first longest word.
# Code: Use lower(), split(), count(), len(), and a for loop.

sentence = "apple mango banana"
words = sentence.lower().split()
longest = ""
for word in words:
    if words.count(word) == 1:
        if len(word) > len(longest):
            longest = word
print(longest)
```
