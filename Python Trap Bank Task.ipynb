```python
# Q35
# [[1,99],[3,4]]
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
# copy.copy() creates a shallow copy. The outer list is copied, but the inner lists are shared. Therefore, changing b[0][1] also changes a
```

```python
# Q36
# [[1, 2],[3, 4]]
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
#  deepcopy() creates completely independent copies of the outer and inner lists. Therefore, changing b does not affect 
```

```python
# Q37
# [[10],[20, 30]]
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
# .copy() creates a shallow copy. The inner lists are shared between a and b, so adding 30 to the inner list affects b
```

```python
# Q38
# [[10],[20]]
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
# deepcopy() creates independent inner lists. Therefore, modifying a does not affect b
```

```python
# Q39
# False,True
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
# The outer lists are different objects, so a is b is False. But shallow copying keeps the same inner lists, so a[0] is b[0] is True
```

```python
# Q44
# (1, 2)
def test(*args):
    print(args)

test(1, 2)
# *args collects all positional arguments into a tuple. Therefore, args contains (1, 2)
```

```python
# Q45
# tuple
def test(*args):
    print(type(args))

test("a", "b", "c")
# The *args parameter always stores positional arguments as a tuple
```

```python
# Q46
# {'name':'Ram','age':25}
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)
# **kwargs collects keyword arguments into a dictionary. The keys are name and age
```

```python
# Q47
# <class 'dict'>
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)
# **kwargs always stores keyword arguments in a dictionary
```

```python
# Q49
# 10
def test():
    return 10

x = test()
print(x)
# return 10 sends the value 10 back to the caller. Therefore, x gets the value 10, which is then printed
```

```python
# Q50
# Python,100
def test():
    print("Python")
    return 100

x = test()
print(x)
# When test() is called, print("Python") executes first, so Python is displayed. Then return 100 sends 100 back to x. Finally, print(x) displays 100
```
