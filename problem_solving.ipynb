Q1.Frequency of Each Word
 1.Split the sentence into individual words.
 2.Create an empty dictionary.
 3.Count each word and store its frequency in the dictionary.
4.Print the dictionary.
expected:  {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}


```python

def word_frequency(sentence):
    frequency = {}
    for word in sentence.split():
        frequency[word] = frequency.get(word, 0) + 1
    return frequency
print(word_frequency("python is easy and python is powerful"))
print(word_frequency("cat dog cat"))
```

Q2. Frequency of Each Character
Pseudocode
1. Create an empty dictionary.
2. Traverse each character in the word.
3. Increase the character's count if it exists; otherwise, set its count to 1.
4. Print the dictionary.
expected: {'b': 1, 'a': 3, 'n': 2}

```python

def character_frequency(word):
    frequency = {}
    for char in word:
        frequency[char] = frequency.get(char, 0) + 1
    return frequency
print(character_frequency("banana"))
print(character_frequency("hello"))
```

Q3. First Non-Repeating Character
Pseudocode
1. Count the frequency of every character.
2. Traverse the word from left to right.
3. Return the first character with frequency 1.
4. If no such character exists, return None.
expected: w

```python
def first_non_repeating(word):
    frequency = character_frequency(word)
    for char in word:
        if frequency[char] == 1:
            return char
    return None
print(first_non_repeating("swiss"))
print(first_non_repeating("aabb"))
```

Q4. Check Whether Two Strings Are Anagrams
Pseudocode
1. Convert both strings to lowercase and remove spaces.
2. Count the characters in both strings.
3. Compare the frequency dictionaries.
4. Return True if they match; otherwise return False.
expected: True

```python

def are_anagrams(s1, s2):
    s1 = "".join(s1.lower().split())
    s2 = "".join(s2.lower().split())
    return character_frequency(s1) == character_frequency(s2)
print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))
```

Q5. First Repeating Character
Pseudocode
1. Create an empty set to store visited characters.
2. Traverse the string from left to right.
3. If a character is already visited, return it.
4. Otherwise, add it to the set. Return None if no character repeats.
expected: r

```python

def first_repeating_char(text):
    seen = set()
    for char in text:
        if char in seen:
            return char
        seen.add(char)
    return None

print(first_repeating_char("programming"))
print(first_repeating_char("abcde"))
```

Q6. Move All Zeros to the End
Pseudocode
1. Collect all non-zero numbers in their original order.
2. Count the zeros in the list.
3. Append all zeros to the non-zero numbers.
4. Return the new list.
expected: [3,1,2,4,0,0,0]

```python

def move_zeros(nums):
    non_zeros = [num for num in nums if num != 0]
    zeros = [0] * (len(nums) - len(non_zeros))
    return non_zeros + zeros
print(move_zeros([0, 3, 0, 1, 2, 0, 4]))
print(move_zeros([0, 0, 5, 6]))
```

Q7. Second Highest Distinct Number
Pseudocode
1. Remove duplicate numbers using a set.
2. Sort the distinct numbers in descending order.
3. If at least two distinct numbers exist, return the second number.
4. Otherwise, return None.
expected: 30

```python

def second_highest(nums):
    unique_nums = sorted(set(nums), reverse=True)
    if len(unique_nums) < 2:
        return None
    return unique_nums[1]
print(second_highest([10, 20, 40, 30, 40, 20]))
print(second_highest([5, 5, 5]))

```

Q8. Remove Duplicates Without Changing Order
Pseudocode
1. Create an empty result list and a set.
2. Traverse the original list.
3. If an element has not been seen, add it to both collections.
4. Return the result list.
expected: [4,2,5,7]

```python

def remove_duplicates(nums):
    result = []
    seen = set()

    for num in nums:
        if num not in seen:
            result.append(num)
            seen.add(num)

    return result
print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))
print(remove_duplicates([1, 1, 2, 3, 2]))
```

Q9. Find the Missing Number from 1 to N
Pseudocode
1. Calculate the expected sum using the formula \(N(N+1)/2\).
2. Calculate the sum of the given list.
3. Subtract the list sum from the expected sum.
4. Return the difference.
expected: 3

```python

def missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(missing_number([1, 2, 4, 5, 6], 6))
print(missing_number([1, 3, 4], 4))

```

Q10. Find Two Numbers Whose Sum Equals a Target
Pseudocode
1. Traverse the list from left to right.
2. Treat each number as the second element of a possible pair.
3. Check earlier numbers in their original order.
4. Return the first valid pair, or an empty list if no pair exists.
expected: [2,7]

```python

def two_sum_first_pair(nums, target):
    for j in range(len(nums)):
        for i in range(j):
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
    return []

print(two_sum_first_pair([2, 7, 11, 15], 9))
print(two_sum_first_pair([1, 2, 4], 20))
```

Q11. Find the Number Closest to Zero
Pseudocode
1. Compare the absolute values of all numbers.
2. Select the number with the smallest absolute value.
3. If two numbers are equally close to zero, choose the positive number.
4. Return the selected number.
expected: 2

```python

def closest_to_zero(nums):
    if not nums:
        return None
    return min(nums, key=lambda num: (abs(num), -num))

print(closest_to_zero([-8, -3, 2, 5, -2]))
print(closest_to_zero([-4, -1, 3]))
```

Q12. Find Common Elements Between Two Lists
Pseudocode
1. Convert the second list into a set.
2. Traverse the first list in its original order.
3. Select elements that occur in the second list.
4. Add each common element only once and return the result.
expected: [3,4,5]

```python

def common_elements(list1, list2):
    set2 = set(list2)
    result = []
    seen = set()

    for item in list1:
        if item in set2 and item not in seen:
            result.append(item)
            seen.add(item)

    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print(common_elements([3, 1, 3, 2], [2, 3]))

```

Q13. Rotate a List to the Right by K Positions
Pseudocode
1. If the list is empty, return an empty list.
2. Calculate the effective rotations using the remainder of K divided by the list length.
3. Split the list into the last K elements and the remaining elements.
4. Join the last part before the first part.
expected: [4,5,1,2,3]

```python

def rotate_right(nums, k):
    if not nums:
        return []

    k = k % len(nums)

    if k == 0:
        return nums[:]

    return nums[-k:] + nums[:-k]

print(rotate_right([1, 2, 3, 4, 5], 2))
print(rotate_right([1, 2, 3], 4))
```

Q14. Maximum Sum of Two Consecutive Elements
Pseudocode
1. Calculate the sum of the first two elements.
2. Traverse the remaining consecutive pairs.
3. Calculate each pair's sum and update the maximum if necessary.
4. Return the maximum sum.
expected: 14

```python

def max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]

    for i in range(1, len(nums) - 1):
        pair_sum = nums[i] + nums[i + 1]

        if pair_sum > max_sum:
            max_sum = pair_sum

    return max_sum

print(max_consecutive_sum([4, 7, 2, 9, 5]))
print(max_consecutive_sum([-5, -2, -8]))
```

Q15. Find All Unique Pairs Whose Sum Equals a Target
Pseudocode
1. Check every pair of different positions in the list.
2. If the two numbers add up to the target, arrange them in ascending order.
3. Store each unique pair only once.
4. Sort the pairs by their first element and return the result.
expected: [[1,8],[2,7],[4,5]]

```python

def all_pairs_sum(nums, target):
    pairs = set()

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = tuple(sorted((nums[i], nums[j])))
                pairs.add(pair)

    return [list(pair) for pair in sorted(pairs)]

print(all_pairs_sum([2, 4, 3, 5, 7, 8, 1], 9))
print(all_pairs_sum([1, 2, 3], 10))
```
